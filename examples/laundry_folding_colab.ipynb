## Gymnasium contract

Every action contains eight normalized velocity commands: three joints and one symmetric two-finger gripper for each arm. Positive gripper values open the fingers; negative values close them. The observation includes both robots, every towel vertex and task progress, illustrating why deformable-object learning is difficult.

```python
import gymnasium as gym
import kaist_rl_lab

env = gym.make("kaist-or/LaundryFoldingEnv-v0")
observation, info = env.reset(seed=7001)
print("Action shape:", env.action_space.shape)
for index, name in enumerate(env.unwrapped.ACTION_NAMES):
    print(f"  {index}: {name}")
print("Observation size:", observation.size)
print("Initial stage:", info["stage"])
env.close()
```

## What to notice

- Commands stay active over a discrete decision interval while the towel moves through smaller physics steps.
- Closing a gripper captures cloth only when the towel is between its two fingers.
- The task begins in **Straighten** and unlocks **Fold** only after sustained bimanual tension and straightness.
- A random policy almost never coordinates grasping, tension, and folding. Human demonstrations therefore provide a useful starting point for behavior cloning before reinforcement-learning fine-tuning.
