## Using the controls

The controls stay below the scene so both robots, the cup, and the pot remain visible.
The teal group controls the **cup arm**; the blue group controls the **pot arm**.

- **↺ / ↻**: rotate a joint counterclockwise / clockwise. A direction stays active until changed.
- **■ Hold**: stop one joint. Several other joints can keep moving together.
- **Pause time / Resume time**: freeze or resume the simulation, preserving your selected joint commands.
- **Stop all motors**: hold all six joints. This preserves whether time is running or paused.
- **Reset + start**: begin a fresh attempt with all joints held and time running.

The demo uses normal speed and has no time limit. A joint takes about ten simulated seconds to rotate 90°, unless it reaches a mechanical or contact limit.
