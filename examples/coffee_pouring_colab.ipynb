## Plan, pour, submit

- **↺ / ↻** select a joint's direction; **■ Hold** stops that joint. Commands stay selected until changed.
- **Pause time / Resume time** preserves all selected commands while freezing or resuming the simulation.
- **Stop all motors** holds all joints without changing the pause state.
- **Reset + start** starts a new running attempt with all joints held.

When your attempt is ready, open **Save your demonstration** below the scene. Enter the participant code your instructor requested (optional), then press **Submit trajectory**. Submission ends the attempt. Wait for a receipt confirming the number of transitions received before resetting.
