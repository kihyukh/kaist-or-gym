## Controls

- Joint controls are docked below the scene: cup arm on the left, pot arm on the right. Each joint has counterclockwise, hold, and clockwise buttons.
- A direction remains latched, so several joints can rotate at the same time.
- At full command, each joint takes about ten simulated seconds to rotate 90 degrees (unless it reaches a mechanical stop).
- The toolbar above the scene shows the current step and simulated time.
- **Pause time** freezes the environment without clearing latched commands; the button changes to **Resume time** while paused.
- **Reset + start** starts a fresh episode with all joints held.
- **Stop all motors** sends a zero action but does not stop the episode clock.
- The demo uses a 700 mL target, normal playback speed, and unlimited practice time. Training environments are capped at 330 steps by default.
