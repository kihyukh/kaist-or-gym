## Controls

- Each joint has counterclockwise, stop, and clockwise buttons.
- A direction remains latched, so several joints can rotate at the same time.
- At full command, each joint takes about ten simulated seconds to rotate 90 degrees (unless it reaches a mechanical stop).
- The large timeline above the scene shows the current step and simulated time.
- **Pause / resume time** freezes the environment without clearing latched commands.
- **Time speed** changes wall-clock playback speed without changing the environment time step or dynamics.
- **Stop all motors** sends a zero action but does not stop the episode clock.
- Human practice has no time limit by default. Enable **Cap episode on reset** and choose a maximum step count for a finite attempt. Training environments are capped at 330 steps by default.
- **Finish** stops the timer for that episode and marks the last recorded transition as manually truncated.
- **Export human demonstration** downloads an `.npz` file with observations, actions, rewards, next observations, episode flags, timing metadata, and a manual-finish flag.
